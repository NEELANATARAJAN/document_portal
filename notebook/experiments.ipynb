## 1. Data Ingestion

1. a. Load PDF data

```python
from langchain.document_loaders import PyPDFLoader
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter
```

```python
import os
```

```python
file_path=os.path.join(os.getcwd(), "data", "sample.pdf")
```

```python
loader=PyPDFLoader(file_path=file_path)
```

```python
document=loader.load()
```

```python
len(document)
```

```python
document
```

1. b. Convert the text data into Chunks

```python
# chunk size and overlap are hyperparameters
text_splitter=RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=150,
    length_function=len
)
```

```python
docs=text_splitter.split_documents(document)
```

```python
len(docs)
```

```python
docs
```

```python
docs[0].metadata
```

```python
docs[0].page_content
```

1. c. Convert the chunks into embedding vectors

```python
embedding_model.embed_documents(docs[0].page_content)
```

```python
from langchain.vectorstores import FAISS
```

1. d. Store the embedded vectors into Vector DB

```python
vectorstore=FAISS.from_documents(docs, embedding_model)
```

#1. D. Data Retrieval

```python
relevant_doc=vectorstore.similarity_search("llama2 finetuning benchmark experiments", k=10)
```

```python
relevant_doc
```

### Retriever

```python
retriever=vectorstore.as_retriever(search_kwargs={'k' : 10})
```

```python
retriever.invoke("llama2 finetuning benchmark experiments.")
```

## Question: User Question
## Context: Based on the question retrieving the info from the vector database

### Prompt

```python
prompt_template="""
Answer the question based on the context provided below.
If the context does not contain sufficient information, respond with:
"I do not have enough information about this."

Context: {context}

Question: {question}

Answer: """
```

```python
from langchain.prompts import PromptTemplate
```

```python
prompt=PromptTemplate(
    template=prompt_template,
    input_variables=["context","question"]
)
```

```python
prompt
```

### Output parser

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
parser=StrOutputParser()
```

### Page Content from DB

```python
def format_docs(docs):
    return "\n\n".join([doc.page_content for doc in docs])
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
rag_chain = (
    {"context":retriever | format_docs, "question": RunnablePassthrough()}
    | prompt
    | model
    | parser
)
```

```python
rag_chain.invoke("tell me about llama2 finetuning benchmark experiments?")
```
